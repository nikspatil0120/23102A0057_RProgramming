### For Loop (1 to 5)


```r
for( i in 1:5 )
{
    print( i^2 )
}
```

### For Loop (Specific Vector)


```r
for(i in c(2,4,6,7))
{
    print(i^2)
}
```

### Function with For Loop and Conditional


```r
x=c(2,4,6.8,10,12)
excount=function(x){
  count=0
  for(xval in x)
  {
    if(xval/2>3)
    count=count+1
  }
print(count)
}
excount(x)
```

### Nested For Loops


```r
child = c("child1","child2","child3")
sweet = c("sweet1","sweet2","sweet3")
for(x in child)
{
  for(y in child)
  print(paste(x,y))
}
```

### For Loop with `break` Statement


```r
drink = c("coffee","lemonade","tea","juice")
for(x in drink)
{
  if(x=="tea")
  break
  print(x)
}
```

### For Loop with `next` Statement


```r
drink = c("coffee","lemonade","tea","juice")
for(x in drink)
{
  if(x=="tea")
  next
  print(x)
}
```

### While Loop


```r
i=1
while(i<10)
{
  print(i^2)
  i=i+2
}
```

### Function with User Input and While Loop


```r
sumfunction=function(){
  sum=0
  number=as.integer(readline(prompt="Please select any number less than 25: "))
  while(number<=25){
    sum=sum+number
    number=number+1
  }
  print(paste("The sum of numbers received from the While Loop: ",sum))
}
```

```r
sumfunction()
```

### Repeat Loop with `break`


```r
i=1
repeat{
  print(i^2)
  i=i+2
  if(i>10)
  break
}
```

### Repeat Loop with `next` and `break`


```r
i=1
repeat{
  i=i+1
  if(i<10)next
  print(i^2)
  if(i>13)break
}
```

### Simple Function `abc(x)`


```r
abc=function(x){
  x^2
}
abc(3)
abc(6)
abc(9)
```

### Function `abc(x, y)` with Two Arguments


```r
abc=function(x,y){
  x^2+y^2
}
abc(3,4)
abc(10,10)
abc(-2,-3)
```

### Function `abc(x)` with Trigonometric Identity


```r
abc=function(x){
  sin(x)^2+cos(x)^2+x
}
abc(9)
abc(99)
abc(-15)
```

### Function Containing a Loop


```r
ab=function()
{
  for(i in 1:3)
  {
    print(i^3)
  }
}
ab()
```
