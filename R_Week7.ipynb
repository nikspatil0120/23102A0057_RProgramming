##Repeat

The `rep()` function in R is used for replicating elements of vectors. `rep(3.5, times=10)` replicates the number 3.5 ten times.

```r
rep(3.5,times=10)
```

This replicates the vector `1:4` (which is `c(1, 2, 3, 4)`) two times.

```r
rep(1:4,2)
```

This defines a vector `x` as a sequence from 1 to 4 and then displays it.

```r
x=1:4
x
```

This replicates the vector `x` (which is `c(1, 2, 3, 4)`) three times.

```r
rep(x,times=3)
```

This replicates each element of vector `x` individually three times before moving to the next element.

```r
rep(x,each=3)
```

This replicates each element of the sequence `1:4` two times.

```r
rep(1:4,each=2)
```

This command combines `each` and `times`. It replicates each element of `1:4` two times, and then repeats this entire sequence three times.

```r
rep(1:4,each=2,times=3)
```

This command is similar to the previous one, but with `times` and `each` arguments swapped. It replicates the entire vector `1:4` three times, and then each element within that resulting sequence is replicated two times.

```r
rep(1:4,times=3,each=2)
```

This replicates each element of `1:4` a different number of times, as specified by the vector `2:5`. So, 1 is repeated 2 times, 2 is repeated 3 times, 3 is repeated 4 times, and 4 is repeated 5 times.

```r
rep(1:4,2:5)
```

This defines a sequence `ans` starting from 2, ending at 8, with a step of 2, and then displays `ans`.

```r
ans=seq(from=2,to=8,by=2)
ans
```

This replicates each element of `1:4` by the corresponding count in the `ans` vector. So, 1 is repeated 2 times, 2 is repeated 4 times, 3 is repeated 6 times, and 4 is repeated 8 times.

```r
rep(1:4,ans)
```

##Mode

The `mode()` function in R returns the *storage mode* of an object, not the statistical mode (most frequent value). For a numeric value like `2.432`, its mode is "numeric".

```r
mode(2.432)
```

For a numeric vector, the `mode()` function returns "numeric".

```r
mode(c(3,4,5,6,7,8))
```

For a character string, the `mode()` function returns "character".

```r
mode("India")
```

For a character vector, the `mode()` function returns "character".

```r
mode(c("India","CANADA"))
```

For a factor object (categorical variable), the `mode()` function returns "numeric" because factors are internally stored as integer codes.

```r
mode(factor(c("UP","MP")))
```

For a list, the `mode()` function returns "list".

```r
mode(list("INDIA","USA"))
```

For a data frame, the `mode()` function returns "list" because a data frame is essentially a list of vectors of equal length.

```r
mode(data.frame(x=1:2,y=c("India","USA")))
```

For a function, the `mode()` function returns "function".

```r
mode(print)
```

This creates a list `z1` containing three different types of objects: a character vector, a numeric vector (generated by `rep`), and a matrix. It then displays the list.

```r
z1=list(c("water","juice","lemonade"),rep(1:4,each=2),matrix(data=5:8,nrow=2,ncol=2,byrow=T))
z1
```

This uses double square brackets `[[]]` to extract the *content* of the first element of list `z1`. This extracts the actual character vector `c("water", "juice", "lemonade")`.

```r
z1[[1]]
```

This attempts to extract the second element from a sub-list created by `z1[1]`. `z1[1]` returns a list containing only the first element of `z1`. Trying to index this new list with `[2]` will result in `NULL` if there isn't a second element in the sub-list (which there isn't).

```r
z1[1][2]
```
